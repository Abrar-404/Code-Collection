```python
x = 10; y = 3

sum = x + y
sub = x - y
div = round(x / y, 2)
mult = x * y
rem = x % y

print(f"Sum: {sum} Sub: {sub} Div: {div} Mult: {mult} Rem: {rem}")
```

```python
# conditionals

x = 10; y = 3

greater_than = x > y

greater_than_equal = (10 >= 10)

print(f"Greater than: {greater_than} Greater than or equal: {greater_than_equal}")


less_than = x < y

less_than_equal = (10 <= 10)

print(f"Less than: {less_than} Less than or equal: {less_than_equal}")

equal = x == y
not_equal = x != y

print(f"Equal: {equal} Not equal: {not_equal}")
```
