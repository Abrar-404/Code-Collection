## Logical Operators

```python
x = 10; y = 5; z = 3

result = (x > y) and (x > z)
print(result)  # Output: True
```

```python
x = 10; y = 5; z = 15

result = (x > y) and (x > z)
print(result)  # Output: False
```

```python
x = 10; y = 5; z = 15

result = (x > y) or (x > z)
print(result)  # Output: True
```

```python
x = 10; y = 20; z = 15

result = (x > y) or (x > z)
print(result)  # Output: False
```
