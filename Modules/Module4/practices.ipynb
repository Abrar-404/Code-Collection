## Smart Light Pilot

```python
brightness, threshold = input().split()

if float(brightness) >= float(threshold):
  print('ON')
else:
  print('OFF')
```

## Sensor Calibrator — Min-Max Scaling

```python
x, min_v, max_v = input().split()

norm = (float(x) - float(min_v)) / (float(max_v) - float(min_v))
print(f"{norm:.2f}")
```

## Mini Loss Tracker

```python
n = int(input())
target = float(input())

total = 0
for i in range(n):
  total += float(input())

avg = total / n

if avg <= target:
  print('PASS')
else:  
  print('RETRY')
```

## Majority Vote Gate

```python
n = int(input())

yes_vote = 0
no_vote = 0

for i in range(n):
  vote = input()
  if (vote == 'YES'):
    yes_vote += 1
  else:
    no_vote += 1

if yes_vote >= no_vote:
    print('ACCEPT')
else:
    print('REJECT')
```

## AI Mood Detector

```python
happy_words = ['happy', 'joy', 'smile']
sad_words = ['sad', 'cry', 'angry']

sentence = input().lower()
words = sentence.split()

if(any(word in happy_words for word in words)):
    print('Happy Mood')
elif(any(word in sad_words for word in words)):
    print('Sad Mood')
else:    
   print('Neutral Mood')
```

## AI Brightness Analyzer

```python
x = [int(a) for a in input().split()]

avg = sum(x) / len(x)

if avg < 85:
  print('Dark Image')
elif avg <= 170:
  print('Normal Image')
else:
  print('Bright Image')
```

## AI Bias Detector

```python
x = input().split()

total = len(x)

a_count = x.count('A')
b_count = x.count('B')

if(a_count / total > 0.70) or (b_count / total > 0.70):
  print('Biased Model')
else:
  print('Fair Model')
```

## AI Keyword Counter

```python
message = input().split()
keywords = ['ai', 'data', 'model', 'learn', 'train', 'neural']

count = 0
for word in keywords:
  if word in message:
    count += 1

if count >= 2:
  print('AI Detected')
else:
  print("Not AI Related")
```
