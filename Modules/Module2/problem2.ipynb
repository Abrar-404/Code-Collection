## Even, Odd, Positive and Negative

```python
n = int(input())

numbers = input().split()

pos = 0; neg = 0; odd = 0; even = 0

for i in range(n):
 
  x = int(numbers[i])
  # Positive and Negative
  if(x > 0):
    pos += 1
  elif(x < 0):
    neg += 1

# Odd and Even
  if(x % 2 != 0):
    odd += 1
  elif(x % 2 == 0):
    even += 1

print("Even:", even)
print("Odd:", odd)
print("Positive:", pos)
print("Negative:", neg)
```
