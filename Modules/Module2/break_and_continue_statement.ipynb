## Continue

```python
sum = 0

for i in range(1, 11):
    print('Processing number....')

    if(i % 2 == 0):
        continue

    sum += i
    print(f"Sum after adding {i}: {sum}")
```

## Break

```python
accuracy = 95

for i in range(20):
  accuracy += 1
  print(f"Accuracy after iteration {i}: {accuracy}%")
  if accuracy == 100:
    break
```
