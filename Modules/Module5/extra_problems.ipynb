## Count Letters

```python
letters = input()

count = {}

for words in letters:
  count[words] = count.get(words, 0) + 1

for letter in sorted(count):
    print(letter, ":", count[letter])
```

## No Idea! - Hackerrank

```python
n, m = input().split()
array = input().split()
like = set(input().split())
dislike = set(input().split())

happiness = 0

for x in array:
    if x in like:
        happiness += 1
    elif x in dislike:
        happiness -= 1

print(happiness)
```

## Hashed Tuples - Hackerrank

```python
n = int(input())
t = tuple(int(x) for x in input().split())

result = hash(t)
print(result)
```
