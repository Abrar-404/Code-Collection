```python
Set = { 1 ,2 ,3 , 4 ,1 ,2 } 

if (10 in Set):
    print("3 is present")
else:
    print("not present")

sum = 0 

for element in Set:
    sum += element 

print(sum)
```

```python
Set = {5, 1 , 2 ,4, 3} 

Set.add(10) 
Set.add(0)

print(Set)

Set.pop()
Set.pop()
print(Set)

Set.remove(5) 

print(Set)
```

## Set Methods

```python
a = { 1, 2, 3 } 
b = {1, 2 , 5,6, 4 } 

print( a.union(b) )
print(a.intersection(b))
print(a.isdisjoint(b)) 
print(a.issubset(b)) 
```
