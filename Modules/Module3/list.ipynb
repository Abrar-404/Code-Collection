## Lists

```python
numbers = [100,20,30,40,50,100,100]

# access
print(numbers[2])

numbers[2] = 43

print(numbers[2])

new_list = numbers[0:3]

print(new_list)
```

```python

## adding a item 
numbers.append(60)
print(numbers)


## insertion at any position 
numbers.insert(0,35)
print(numbers)


## deleting the last value 
numbers.pop()

print(numbers)

## deleting a specific values 
numbers.remove(100)

print(numbers)
```

```python
numbers.sort()
print(numbers)

numbers.reverse()
print(numbers)
```
