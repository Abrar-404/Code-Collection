## p1. Lowest Number

```python
n = int(input())
x = [int(x) for x in input().split()]

min_value = min(x)
value_position = x.index(min_value) + 1

print(min_value, value_position)
```

## p2. Replace MinMax

```python
n = int(input())
x = [int(x) for x in input().split()]

min_value = min(x)
max_value = max(x)

min_position = x.index(min_value)
max_position = x.index(max_value)

x[min_position], x[max_position] = x[max_position], x[min_position]
print(*x)
```

## p3. Sum Digits

```python
n = int(input())
x = input()

sum_digits = sum([int(a) for a in x])
print(sum_digits)
```

## p4. Good or Bad

```python
t = int(input())

for i in range(t):
    s = input()
    if "010" in s or "101" in s:
        print("Good")
    else:
        print("Bad")
```
